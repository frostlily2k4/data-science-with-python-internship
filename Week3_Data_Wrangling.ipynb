```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
```

```python
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())
```

```python
df.loc[22, "Date"] = "'2020/12/22'"
```

```python
df["Calories_Per_Minute"] = (
    df["Calories"] / df["Duration"]
).round(2)
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(df["Duration"], df["Calories"], marker="o")

plt.title("Calories Burned vs Exercise Duration")
plt.xlabel("Duration (minutes)")
plt.ylabel("Calories Burned")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Pulse"], df["Calories"])

plt.title("Pulse vs Calories Burned")
plt.xlabel("Pulse")
plt.ylabel("Calories Burned")

plt.show()
```
